```python
INSTRUCTIONS = '''
You are an expert financial analyst who specializes in analyzing data related to expenses. 
Your task is to come up with potential questions (ranging from easy to difficult) 
a user may ask you based on the schema below, as well as the valid SQLITE SQL queries. 

The earliest entries in the table are from 2025. The existing categories are the following:

- Bills
- Subscriptions
- Entertainment
- Food & Drink
- Groceries
- Health & Wellbeing
- Other
- Shopping
- Transport
- Travel
- Gifts

SCHEMA:
CREATE TABLE IF NOT EXISTS expenses (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    timestamp TEXT NOT NULL,
                    purchase_date DATE NOT NULL,
                    item TEXT,
                    amount REAL NOT NULL,
                    category TEXT NOT NULL
                    )
'''
```

```python
import json
from google import genai
from dotenv import load_dotenv
from tqdm import tqdm
load_dotenv()
llm_client = genai.Client()
```

```python
from rag_helper import RAGBase, GoldenDataSet

assistant = RAGBase(llm_client=llm_client, instructions=INSTRUCTIONS, model='gemini-3.1-pro-preview')
```

```python
prompt = "Generate 50 different questions and queries of various degrees of difficulty based on the provided schema"
```

```python
response = assistant.llm_structured_response(prompt, schema=GoldenDataSet.model_json_schema())
```

```python
result = json.loads(response.output_text)
```

```python
result_list = result["questions"]
```

```python
import pandas as pd
```

```python
result_df = pd.DataFrame(result_list)
```

```python
result_df.to_csv("data/ground_truth_sql.csv")
```
